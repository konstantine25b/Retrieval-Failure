# RealMistake Enriched Dataset

Full view of `data/realmistake_full_enriched.csv` with model architecture, inference defaults, and benchmark scores joined by `llm_model`.

Generate the file with:

```bash
python scripts/enrich_realmistake_full.py
```

````python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_PATH = PROJECT_ROOT / "data" / "realmistake_full_enriched.csv"

df = pd.read_csv(DATA_PATH)
df.head()
````

## Overview

````python
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print()
print(list(df.columns))
print()
df.info()
````

## Model feature profiles (one row per model)

````python
feature_cols = [c for c in df.columns if c not in ("question", "llm_model", "error")]
model_profiles = df.groupby("llm_model", as_index=False)[feature_cols].first()
model_profiles.T
````

## Label balance by model

````python
pd.crosstab(df["llm_model"], df["error"], margins=True)
````

## Full dataset (all columns)

Questions are truncated for display; all feature columns are shown in full.

````python
display_df = df.copy()
display_df["question"] = display_df["question"].str.slice(0, 120) + "..."

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 300)
pd.set_option("display.max_colwidth", 40)

display_df
````
