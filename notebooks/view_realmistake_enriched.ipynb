# RealMistake Enriched Dataset (Full View)

All columns in `data/realmistake_full_enriched.csv`: original labels, model features, and question-based features.

Generate the file with:

````python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_PATH = PROJECT_ROOT / "data" / "realmistake_full_enriched.csv"

df = pd.read_csv(DATA_PATH)
df.head(3)
````

## Overview

````python
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print()
print(list(df.columns))
print()
df.info()
````

## Question-based feature summary

````python
question_cols = [
    "question_length_words",
    "question_length_chars",
    "question_complexity_score",
    "has_few_shot_examples",
    "prompt_contains_system_instructions",
    "question_category",
    "is_ambiguous",
    "contains_negation",
    "context_token_count",
]
df[question_cols].describe(include="all").T
````

````python
pd.crosstab(df["question_category"], df["error"], margins=True)
````

## Model feature profiles

````python
model_cols = [c for c in df.columns if c not in ("question", "error") and c not in question_cols]
df.groupby("llm_model", as_index=False)[model_cols].first().T
````
